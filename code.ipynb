## step 4 : Function Engineering

#### The is_investigational variable was evaluated for feature engineering; however, all adverse-event records corresponded to approved drugs (False). Therefore, an approval-status feature was not created because it would have no variability and would not contribute to the analysis.

```python
df["is_investigational"].value_counts(dropna=False)
```

```python
# 1-drop the "is_investigational" column from the dataframe
df.drop(columns=["is_investigational"], inplace=True)
```

```python
# 2-Extract the year from the "fda_first_approval_date" column

df["approval_year"] = df["fda_first_approval_date"].dt.year

df["approval_year"].unique()
```

```python
# 3-Calculate the number of years since approval
df["years_since_approval"] = (
    (df["receive_date"] - df["fda_first_approval_date"]).dt.days / 365.25
).astype(int)
df["years_since_approval"].unique()
```

```python
# To investegate which drug have report before thier approval date

df[df["years_since_approval"] < 0][
    ["generic_name", "fda_first_approval_date", "receive_date", "years_since_approval"]
].drop_duplicates().sort_values("years_since_approval")
```

```python
df.loc[
    df["years_since_approval"] < 0,
    "generic_name"
].value_counts()
```

```python
# 4-Assign severity levels based on the "serious_outcome" column
# 0 = No specified serious outcome
# 1 = Disabling
# 2 = Hospitalization
# 3 = Life-threatening
# 4 = Death

def get_severity(row):
    if row["seriousness_death"] == True:
        return 4
    elif row["seriousness_lifethreatening"] == True:
        return 3
    elif row["seriousness_hospitalization"] == True:
        return 2
    elif row["seriousness_disabling"] == True:
        return 1
    else:
        return 0

df["severity_level"] = df.apply(get_severity, axis=1)
```

```python
# get the severety of each report

df_severity = (
    df.groupby("safetyreportid", as_index=False)["severity_level"]
      .max())
df_severity
```
