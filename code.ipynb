```python
import pandas as pd
import numpy as np
import plotly.express as px
```

## step 1: Load the data

```python
df_adverse_events = pd.read_csv('adverse_events.csv')
df_drugs = pd.read_csv("drugs_overview.csv")
df_prices = pd.read_csv("stock_prices.csv")
df_adverse_events.head()
```

```python
# to load drugs overview data

df = df_adverse_events.merge(df_drugs, on="generic_name", how="left")
df.head()
```

```python
# to show all columns in the dataframe

pd.set_option('display.max_columns', None)
```

## step 2 : Data description / understanding

| Column Name | Description |
|---|---|
| `safetyreportid` | Unique identifier for the safety report. |
| `generic_name` | Generic name of the reported drug. |
| `brand_queried` | Brand name associated with the reported drug. |
| `receive_date` | Date when the adverse-event report was received. |
| `country` | Country associated with the adverse-event report. |
| `serious` | Indicates whether the reported event was classified as serious. |
| `seriousness_death` | Indicates whether the event resulted in death. |
| `seriousness_lifethreatening` | Indicates whether the event was considered life-threatening. |
| `seriousness_hospitalization` | Indicates whether the event resulted in or required hospitalization. |
| `seriousness_disabling` | Indicates whether the event resulted in disability or incapacity. |
| `patient_age` | Age of the patient at the time of the report. |
| `patient_age_unit` | Unit used to report the patient's age. |
| `patient_sex` | Sex of the patient. |
| `patient_weight_kg` | Patient body weight in kilograms. |
| `reaction` | Reported adverse reaction or event. |
| `reaction_outcome` | Outcome of the reported reaction. |
| `brand_names` | Brand names associated with the drug. |
| `manufacturer` | Pharmaceutical company that manufactures or develops the drug. |
| `indication` | Medical condition(s) or purpose(s) for which the drug is used or investigated. |
| `fda_first_approval_date` | Date when the drug first received FDA approval. |
| `is_investigational` | Indicates whether the drug is investigational rather than approved. |
| `drug_class` | Pharmacological class or type of the drug. |

```python
# change the patient_age_unit value from "Decade" to "Unknown" for all rows where patient_age_unit is "Decade"
df.loc[df["patient_age_unit"] == "Decade",["patient_age_unit"]] = "Unknown"
df['patient_age_unit'].value_counts()
```
